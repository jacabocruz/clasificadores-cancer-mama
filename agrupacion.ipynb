# Caso
Un centro de innovación en salud digital está desarrollando un sistema de apoyo a la decisión clínica para mejorar la velocidad y exactitud en el diagnóstico temprano del cáncer de mama. Actualmente, los especialistas analizan cuantitativamente imágenes digitalizadas tomadas a partir de una aspiración con aguja fina (FNA) de masas mamarias, identificando características geométricas y morfológicas de los núcleos celulares.

# Preparación de los datos

## Importacion de pandas y data set

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
```

## Carga del dataset en un Dataframe

```python
data = load_breast_cancer(as_frame=True)
x = data.data.copy()
```

## Se convierte la clase positiva en "Maligno"

```python
y = (data.target == 0).astype(int)  

print(x.shape)
print(y.value_counts())
```

## Division de datos de entrenamiento y pruebas

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

## Función común de evaluación

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, confusion_matrix)

def evaluar_modelo (nombre,modelo,X_eval, y_eval):
    pred = modelo.predict(X_eval)
    
    resultados = {
        'modelo': nombre,
        'accuracy': accuracy_score(y_eval, pred),
        'precision': precision_score(y_eval, pred, average='weighted'),
        'recall': recall_score(y_eval, pred, average='weighted'),
        'f1': f1_score(y_eval, pred, average='weighted'),
        'confusion_matrix': confusion_matrix(y_eval, pred)
    }
    print(nombre)
    print (confusion_matrix(y_eval, pred))
    return resultados
```

# Implementacion de los Clasificadores 

##  Árbol de decisiones / Decision trees 

```python
from sklearn.tree import DecisionTreeClassifier

modelo_dt = DecisionTreeClassifier(max_depth=4, random_state=42)

modelo_dt.fit(X_train, y_train)
res_dt = evaluar_modelo('Decision Tree', modelo_dt, X_test, y_test)

# Imprimir variables individuales
print(
    f"Accuracy: {res_dt['accuracy']:.4f} | "
    f"Precision: {res_dt['precision']:.4f} | "
    f"Recall: {res_dt['recall']:.4f} | "
    f"F1: {res_dt['f1']:.4f}"
)
```

## Random forest

```python
from sklearn.ensemble import RandomForestClassifier

modelo_rf = RandomForestClassifier(n_estimators=100, max_depth=4, random_state=42)


modelo_rf.fit(X_train, y_train)
res_rf = evaluar_modelo('Random Forest', modelo_rf, X_test, y_test)
print(
    f"Accuracy: {res_rf['accuracy']:.4f} | "
    f"Precision: {res_rf['precision']:.4f} | "
    f"Recall: {res_rf['recall']:.4f} | "
    f"F1: {res_rf['f1']:.4f}"
)
```

## Suport Vector Machine (SVM)

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

modelo_svm = make_pipeline(StandardScaler(), SVC(kernel='rbf', C=1.0, gamma='scale'))

modelo_svm.fit(X_train, y_train)
res_svm = evaluar_modelo('SVM', modelo_svm, X_test, y_test)
print(
    f"Accuracy: {res_svm['accuracy']:.4f} | "
    f"Precision: {res_svm['precision']:.4f} | "
    f"Recall: {res_svm['recall']:.4f} | "
    f"F1: {res_svm['f1']:.4f}"
)
```

## K-Nearest Neighbors (K-NN)

```python
from sklearn.neighbors import KNeighborsClassifier

modelo_knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))

modelo_knn.fit(X_train, y_train)
res_knn = evaluar_modelo('KNN', modelo_knn, X_test, y_test)
print(
    f"Accuracy: {res_knn['accuracy']:.4f} | "
    f"Precision: {res_knn['precision']:.4f} | "
    f"Recall: {res_knn['recall']:.4f} | "
    f"F1: {res_knn['f1']:.4f}"
)
```

## Native Bayes

```python
from sklearn.naive_bayes import GaussianNB
modelo_nb = GaussianNB()

modelo_nb.fit(X_train, y_train)
res_nb = evaluar_modelo('Naive Bayes', modelo_nb, X_test, y_test)
print(
    f"Accuracy: {res_nb['accuracy']:.4f} | "
    f"Precision: {res_nb['precision']:.4f} | "
    f"Recall: {res_nb['recall']:.4f} | "
    f"F1: {res_nb['f1']:.4f}"
)
```
